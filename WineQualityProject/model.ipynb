# Wine Quality , Model Building

Train and compare 7 classification models on the good_quality target.

## 1) Imports & Load cleaned dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

import sys
sys.path.append(".")

from utils import *

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

%matplotlib inline

```

```python
df = pd.read_csv("data/winequality-red_cleaned.csv")

print("Dataset shape:", df.shape)
display(df.head())

```

## 2) Train / Test Split & Scaling

```python
X = df.drop(columns="good_quality")
y = df["good_quality"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("Train shape:", X_train.shape)
print("Test shape :", X_test.shape)
print(f"Train good rate: {y_train.mean():.3f}")
print(f"Test good rate : {y_test.mean():.3f}")

```

```python
scaler = StandardScaler()

X_train_s = scaler.fit_transform(X_train)
X_test_s  = scaler.transform(X_test)
```

## 3) Model 1 : Logistic Regression

```python
lr = LogisticRegression(
    max_iter=2000,
    class_weight="balanced",
    random_state=42
)

lr.fit(X_train_s, y_train)

pred_lr = lr.predict(X_test_s)

m_lr = evaluate_model("Logistic Regression", y_test, pred_lr)

plot_confusion_matrix(y_test, pred_lr, "Logistic Regression")
plt.show()
```

## 4) Model 2 : Decision Tree

```python
dt = DecisionTreeClassifier(
    random_state=42,
    class_weight="balanced"
)

dt.fit(X_train, y_train)

pred_dt = dt.predict(X_test)

m_dt = evaluate_model("Decision Tree", y_test, pred_dt)

plot_confusion_matrix(y_test, pred_dt, "Decision Tree")
plt.show()
```

## 5) Model 3 : Random Forest

```python
rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    class_weight="balanced"
)

rf.fit(X_train, y_train)

pred_rf = rf.predict(X_test)

m_rf = evaluate_model("Random Forest", y_test, pred_rf)

plot_confusion_matrix(y_test, pred_rf, "Random Forest")
plt.show()
```

## 6) Model 4 : KNN , find the best K

```python
k_values = range(1, 21)
scores = []

for k in k_values:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train_s, y_train)
    scores.append(model.score(X_test_s, y_test))

best_k = k_values[np.argmax(scores)]

plt.figure(figsize=(8,4))
plt.plot(k_values, scores, marker="o")
plt.axvline(best_k, linestyle="--", color="red", label=f"Best K = {best_k}")
plt.xlabel("K")
plt.ylabel("Test Accuracy")
plt.title("KNN Performance vs K")
plt.legend()
plt.tight_layout()
plt.show()

```

```python

# Final KNN model

knn = KNeighborsClassifier(n_neighbors=best_k)

knn.fit(X_train_s, y_train)

pred_knn = knn.predict(X_test_s)

m_knn = evaluate_model(f"KNN (K={best_k})", y_test, pred_knn)

plot_confusion_matrix(y_test, pred_knn, f"KNN (K={best_k})")
plt.show()
```

## 7) Model 5 : SVM

```python
svm = SVC(
    kernel="rbf",
    probability=True,
    class_weight="balanced",
    random_state=42
)

svm.fit(X_train_s, y_train)

pred_svm = svm.predict(X_test_s)

m_svm = evaluate_model("SVM", y_test, pred_svm)

plot_confusion_matrix(y_test, pred_svm, "SVM")
plt.show()

```

## 8) Model 6 : Gradient Boosting

```python
gb = GradientBoostingClassifier(random_state=42)

gb.fit(X_train, y_train)

pred_gb = gb.predict(X_test)

m_gb = evaluate_model("Gradient Boosting", y_test, pred_gb)

plot_confusion_matrix(y_test, pred_gb, "Gradient Boosting")
plt.show()
```

## 9) Model 7 : Naive Bayes

```python
nb = GaussianNB()

nb.fit(X_train, y_train)

pred_nb = nb.predict(X_test)

m_nb = evaluate_model("Naive Bayes", y_test, pred_nb)

plot_confusion_matrix(y_test, pred_nb, "Naive Bayes")
plt.show()
```

## 12) 5-Fold Cross-Validation

```python
cv_models = {
    "Logistic Regression": (lr, X_train_s),
    "Decision Tree":       (dt, X_train),
    "Random Forest":       (rf, X_train),
    f"KNN (K={best_k})":   (knn, X_train_s),
    "SVM":                 (svm, X_train_s),
    "Gradient Boosting":   (gb, X_train),
    "Naive Bayes":         (nb, X_train)
}

cv_scores = {}

for name, (model, X_input) in cv_models.items():

    print(f"\n{name}")

    scores = cross_validate_model(
        model,
        X_input,
        y_train,
        cv=5
    )

    cv_scores[name] = scores
```

```python
cv_df = pd.DataFrame(cv_scores)

plt.figure(figsize=(12,5))

sns.boxplot(
    data=cv_df,
    palette="Set2"
)

plt.title("5‑Fold Cross‑Validation — F1 Score Distribution")

plt.ylabel("F1 Score")

plt.xticks(rotation=30, ha="right")

plt.tight_layout()
plt.show()
```

## 13) ROC Curves

```python
models_for_roc = {
    "Logistic Regression": lr,
    "Random Forest":       rf,
    f"KNN (K={best_k})":   knn,
    "SVM":                 svm,
    "Gradient Boosting":   gb,
    "Naive Bayes":         nb
}

plt.figure(figsize=(10,7))

for name, model in models_for_roc.items():

    X_input = (
        X_test_s
        if name in {"Logistic Regression", f"KNN (K={best_k})", "SVM"}
        else X_test
    )

    if hasattr(model, "predict_proba"):
        probs = model.predict_proba(X_input)[:,1]
    else:
        probs = model.decision_function(X_input)

    fpr, tpr, _ = roc_curve(y_test, probs)

    model_auc = auc(fpr, tpr)

    plt.plot(
        fpr,
        tpr,
        label=f"{name} (AUC={model_auc:.3f})"
    )

plt.plot([0,1],[0,1],"k--", label="Random")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve Comparison")

plt.legend(loc="lower right")

plt.tight_layout()
plt.show()
```

## 14) Hyperparameter Tuning , Random Forest

```python
param_grid = {

    "n_estimators": [200, 400, 600],

    "max_depth": [None, 10, 20],

    "min_samples_split": [2, 5, 10],

    "min_samples_leaf": [1, 2, 4]

}

grid = GridSearchCV(

    RandomForestClassifier(
        random_state=42,
        class_weight="balanced"
    ),

    param_grid,

    cv=5,

    scoring="f1",

    n_jobs=-1
)

grid.fit(X_train, y_train)

print("Best parameters :", grid.best_params_)
print(f"Best CV F1      : {grid.best_score_:.4f}")
```

```python
rf_tuned = grid.best_estimator_

pred_rf_tuned = rf_tuned.predict(X_test)

m_rf_tuned = evaluate_model(
    "Random Forest (Tuned)",
    y_test,
    pred_rf_tuned
)

plot_confusion_matrix(
    y_test,
    pred_rf_tuned,
    "Random Forest (Tuned)"
)

plt.show()
```

## 17) Final Project Summary
* The goal of this project was to build a binary classification model that predicts whether a wine is of good quality based on its physicochemical properties from the UCI Wine Quality dataset. Because only about 13.6% of the wines fall into the “good quality” category, the dataset is imbalanced. For this reason, the F1 score was used as the main evaluation metric instead of accuracy, since accuracy alone can give misleadingly high results when the positive class is rare.

* Among the tested models, tree‑based ensemble methods such as Random Forest and Gradient Boosting achieved the strongest performance. After hyperparameter tuning using GridSearchCV and applying class_weight=“balanced”, the Random Forest model produced the most reliable results with strong F1 scores and stable cross‑validation performance.

* Feature importance analysis showed that alcohol, sulphates, volatile acidity, total sulfur dioxide, and density were the most influential predictors of wine quality. In general, higher alcohol and sulphates were associated with better wine ratings, while higher volatile acidity tended to reduce wine quality. These patterns are consistent with known chemical characteristics of higher‑quality wines.

* Future improvements could focus on handling class imbalance more effectively using techniques such as SMOTE or probability threshold tuning. Additional work could also include training models on the white wine dataset, applying explainability techniques such as SHAP values, and building a small Streamlit application that allows users to input wine chemistry values and receive a predicted quality score.
