## Gold Price Prediction — Exploratory Data Analysis

This project investigates the factors influencing the price of gold using a daily financial dataset covering the period 2008–2018 (2,290 trading days). The dataset includes the price of the SPDR Gold Shares ETF (GLD) along with several related financial indicators:

SPX — S&P 500 Index
USO — United States Oil Fund ETF
SLV — Silver ETF
EUR/USD — Euro to US Dollar exchange rate
Gold is often considered a safe‑haven asset, and its price can be influenced by movements in equity markets, commodities, and currency exchange rates. This project explores these relationships through exploratory data analysis (EDA) before building predictive models.

The goal is to understand how these financial variables relate to gold prices and evaluate whether they can be used to predict daily GLD prices.

Target Variable:

GLD — continuous variable representing the daily closing price of the Gold ETF.

## 1) Imports & Load Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import math
import warnings

warnings.filterwarnings("ignore")

sns.set_style("whitegrid")
pd.set_option("display.max_columns", None)

%matplotlib inline
```

```python

df = pd.read_csv("data/gold_price_data.csv")

df["Date"] = pd.to_datetime(df["Date"])

df = df.sort_values("Date").reset_index(drop=True)

print("Dataset Shape:", df.shape)

df.head()
```

## 2) Dataset Overview

```python

print("\nDataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes.value_counts())

df.describe(include="all").T.head(20)



```

```python
# Duplicate and Memory Check

print("\nDuplicate Rows:", df.duplicated().sum())

memory_mb = df.memory_usage(deep=True).sum() / 1024**2

print("Memory Usage (MB):", round(memory_mb, 2))

```

## 3) Missing / Invalid Values

```python

missing = df.isnull().sum()

missing_pct = (missing / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    "Missing": missing,
    "Percent": missing_pct
})

missing_df = missing_df[missing_df["Missing"] > 0].sort_values(
    "Missing",
    ascending=False
)

if len(missing_df):
    print("\nMissing Values:")
    print(missing_df)
else:
    print("\nNo missing values found")

```

## 4) Target Variable Distribution

```python

fig, axes = plt.subplots(1, 3, figsize=(16,4))

# Histogram
axes[0].hist(
    df["GLD"].dropna(),
    bins=40,
    color="steelblue",
    edgecolor="black"
)

axes[0].set_title("Distribution — GLD")
axes[0].set_xlabel("GLD")


# Boxplot
sns.boxplot(
    x=df["GLD"].dropna(),
    ax=axes[1],
    color="lightcoral"
)

axes[1].set_title("Boxplot — GLD")


# Log transform distribution
axes[2].hist(
    np.log1p(df["GLD"].dropna()),
    bins=40,
    color="seagreen",
    edgecolor="black"
)

axes[2].set_title("log1p(GLD)")

plt.tight_layout()
plt.show()


print("Skewness :", round(df["GLD"].skew(), 3))
print("Kurtosis :", round(df["GLD"].kurtosis(), 3))

print(df["GLD"].describe().round(2))

```

## 5) Univariate Analysis , Numeric Features

```python

numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

if "GLD" in numeric_cols:
    numeric_cols.remove("GLD")

print(f"\nNumeric features ({len(numeric_cols)}):")

print(numeric_cols[:30])

```

```python

n_show = min(len(numeric_cols), 12)

ncols = 3

nrows = math.ceil(n_show / ncols)

fig, axes = plt.subplots(
    nrows,
    ncols,
    figsize=(15, 3.3 * nrows)
)

axes = np.array(axes).reshape(-1)

for i, col in enumerate(numeric_cols[:n_show]):

    df[col].hist(
        ax=axes[i],
        bins=30,
        color="steelblue",
        edgecolor="black"
    )

    axes[i].set_title(col)

for j in range(n_show, len(axes)):
    axes[j].axis("off")

plt.tight_layout()

plt.show()

```

## 6) Correlation with Target

```python

corrs_all = (
    df[numeric_cols + ["GLD"]]
    .corr()["GLD"]
    .abs()
    .sort_values(ascending=False)
)

top_features = [
    c for c in corrs_all.index
    if c != "GLD"
][:6]

print("\nTop features correlated with GLD:")

print(corrs_all[top_features].round(3))
```

```python
# Scatter Plots: Top Features vs Target

fig, axes = plt.subplots(2, 3, figsize=(16,8))

axes = axes.flatten()

for i, col in enumerate(top_features[:6]):

    axes[i].scatter(
        df[col],
        df["GLD"],
        alpha=0.4,
        s=10,
        color="steelblue"
    )

    axes[i].set_xlabel(col)

    axes[i].set_ylabel("GLD")

    axes[i].set_title(f"{col} vs GLD")

plt.tight_layout()

plt.show()
```

## 7) Correlation Heatmap

```python

top_numeric = corrs_all.head(15).index.tolist()

if "GLD" not in top_numeric:
    top_numeric.append("GLD")

corr_matrix = df[top_numeric].corr()

plt.figure(figsize=(11, 9))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    square=True,
    linewidths=0.5,
    cbar_kws={"shrink": 0.7}
)

plt.title("Top Features — Correlation Matrix")

plt.tight_layout()

plt.show()
```

```python
# Correlation with Target (GLD)

target_corr = (
    df[numeric_cols + ["GLD"]]
    .corr()["GLD"]
    .drop("GLD")
    .sort_values()
)

plt.figure(figsize=(10, max(5, len(target_corr) * 0.25)))

colors = [
    "tomato" if v > 0 else "steelblue"
    for v in target_corr
]

target_corr.plot(
    kind="barh",
    color=colors
)

plt.title("Feature Correlation with GLD")

plt.axvline(0, color="black", linewidth=0.5)

plt.tight_layout()

plt.show()
```

## 8) Time Series Visualization

```python

fig, axes = plt.subplots(2, 1, figsize=(14, 8))


# Gold price over time
axes[0].plot(
    df["Date"],
    df["GLD"],
    color="goldenrod",
    label="GLD (Gold ETF)"
)

axes[0].set_title("GLD Price Over Time")

axes[0].legend()



# Other assets normalized to start value
for col, color in [
    ("SPX", "steelblue"),
    ("USO", "tomato"),
    ("SLV", "silver"),
    ("EUR/USD", "green")
]:

    if col in df.columns:

        normalized = df[col] / df[col].iloc[0] * 100

        axes[1].plot(
            df["Date"],
            normalized,
            label=col,
            color=color
        )

axes[1].set_title("Other Assets — Normalized to 100")

axes[1].legend()

plt.tight_layout()

plt.show()
```

```python
# Scatter Plots GLD vs Major Assets
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes = axes.flatten()

for ax, col in zip(
    axes,
    ["SPX", "USO", "SLV", "EUR/USD"]
):

    if col in df.columns:

        ax.scatter(
            df[col],
            df["GLD"],
            alpha=0.4,
            s=10,
            color="steelblue"
        )

        ax.set_xlabel(col)

        ax.set_ylabel("GLD")

        ax.set_title(f"GLD vs {col}")

plt.tight_layout()

plt.show()
```

## 9) Pair Plot , Top Features

```python

sample = df.sample(
    min(800, len(df)),
    random_state=42
)

sns.pairplot(
    sample[top_features[:4] + ["GLD"]].dropna(),
    diag_kind="kde",
    height=2.2
)

plt.suptitle(
    "Pair Plot — Top 4 Features",
    y=1.02
)

plt.show()

```
