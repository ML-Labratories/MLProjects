# Gold Price Prediction Data Cleaning & Feature Engineering

## 1) Imports & Load Raw Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings; warnings.filterwarnings("ignore")

from utils import load_data, preprocess_data, create_features

sns.set_style("whitegrid")
pd.set_option("display.max_columns", None)

%matplotlib inline
```

```python
df = load_data()

print("Raw shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Date range:", df["Date"].min(), "->", df["Date"].max())

df.head()
```

## 2) Handle Missing / Invalid Values

```python
df_clean = df.copy()

# ensure correct datetime format
df_clean["Date"] = pd.to_datetime(df_clean["Date"])

# sort by date
df_clean = df_clean.sort_values("Date").reset_index(drop=True)

print("After basic cleaning:", df_clean.shape)

```

## 3) Imputation

```python

df_imputed = df_clean.copy()

print("NaN count:", df_imputed.isnull().sum().sum())
```

## 4) Distribution Comparison Before vs After Imputation

```python

numeric_cols = df.select_dtypes(include=[np.number]).columns[:6]

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()

for i, col in enumerate(numeric_cols):

    axes[i].hist(
        df[col].dropna(),
        bins=25,
        alpha=0.5,
        label="Before",
        color="steelblue"
    )

    axes[i].hist(
        df_imputed[col],
        bins=25,
        alpha=0.5,
        label="After",
        color="tomato"
    )

    axes[i].set_title(col)
    axes[i].legend()

plt.tight_layout()
plt.show()

```

## 5) Feature Engineering

```python

df_feat = create_features(df_imputed)

print("After feature engineering:", df_feat.shape)

df_feat.head()
```

```python
# Feature Relationships
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

if "SPX_roll5" in df_feat.columns:
    axes[0].scatter(
        df_feat["SPX_roll5"],
        df_feat["GLD"],
        alpha=0.4,
        s=8,
        color="steelblue"
    )
    axes[0].set_title("SPX rolling-5 vs GLD")

if "SLV_diff" in df_feat.columns:
    axes[1].scatter(
        df_feat["SLV_diff"],
        df_feat["GLD"],
        alpha=0.4,
        s=8,
        color="tomato"
    )
    axes[1].set_title("SLV daily diff vs GLD")

plt.tight_layout()
plt.show()

```

## 6) Encode Categorical Features

```python

cat_cols = df_feat.select_dtypes(include=["object", "category"]).columns.tolist()

print("Categorical columns:", cat_cols)

df_encoded = pd.get_dummies(
    df_feat,
    columns=cat_cols,
    drop_first=True
)

# convert bool to int
bool_cols = df_encoded.select_dtypes(include=["bool"]).columns
df_encoded[bool_cols] = df_encoded[bool_cols].astype(int)

print("Final shape:", df_encoded.shape)

```

## 7) Verify & Save Cleaned Dataset

```python
print("Missing values:", df_encoded.isnull().sum().sum())

```

```python

df_encoded.to_csv(
    "data/gold_price_cleaned.csv",
    index=False
)

print("Saved -> data/gold_price_cleaned.csv")

df_encoded.head()
```
